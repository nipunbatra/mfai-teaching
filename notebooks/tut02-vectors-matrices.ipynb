## 1 · Similarity and projection

Predict the cosine and projection before running the cell.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1., 2., 2.])
y = np.array([2., 0., 1.])
cosine = x @ y / (np.linalg.norm(x) * np.linalg.norm(y))
projection = (x @ y) / (y @ y) * y
print("cosine:", cosine)
print("projection:", projection)
```

## 2 · Watch a matrix move a grid


```python
A = np.array([[2., 1.], [0., 1.]])
t = np.linspace(-2, 2, 9)
fig, ax = plt.subplots(figsize=(6, 5))
for value in t:
    horizontal = np.vstack([np.linspace(-2, 2, 100), np.full(100, value)])
    vertical = np.vstack([np.full(100, value), np.linspace(-2, 2, 100)])
    for line in (horizontal, vertical):
        moved = A @ line
        ax.plot(moved[0], moved[1], color="#2C7A7B", alpha=.45)
ax.set_aspect("equal"); ax.set_title(f"det(A) = {np.linalg.det(A):.1f}"); ax.grid(alpha=.15);
```

## 3 · Least squares as projection

The residual should be orthogonal to every column of $A$.


```python
A = np.array([[1., 0.], [0., 1.], [1., 1.]])
b = np.array([1., 2., 4.])
x_hat, *_ = np.linalg.lstsq(A, b, rcond=None)
r = b - A @ x_hat
print("x_hat:", x_hat)
print("residual:", r)
print("A.T @ residual:", A.T @ r)
```
